```python
def plot_continuous(df, feature):
    df_func = df.copy()
    df_func['churn'] = df_func['churn'].astype('category')

    df_churned = df_func[df_func['churn'] == 1]
    df_retained = df_func[df_func['churn'] == 0]

    fig, (ax1, ax2) = plt.subplots(2,
                                   figsize=(9, 7),
                                   sharex=True,
                                   gridspec_kw={'height_ratios': (.7, 3)})

    colors = ['#1177b4', '#ff7f0e']

    for sub_df, color, label in zip([df_retained, df_churned], colors, ['Retained', 'Churned']):
        sns.histplot(data=sub_df,
                     x=feature,
                     bins=15,
                     color=color,
                     alpha=0.6,
                     edgecolor='firebrick',
                     label=label,
                     kde=False,
                     ax=ax1)

    ax1.legend()

    sns.boxplot(x='churn', data=df_func, palette=colors, ax=ax2)
    ax2.set_ylabel('')
    ax2.set_yticklabels(['Retained', 'Churned'])
    plt.tight_layout()
```

```python
plot_continuous(df,'age')
```

```python
plot_continuous(df,'credit_score')
```

```python
plot_continuous(df,'balance')
```

```python
plot_continuous(df,'estimated_salary')
```
